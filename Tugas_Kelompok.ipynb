```python
from abc import ABC, abstractmethod


class KamarHotel(ABC):

    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = stok
        self.__harga_dasar = harga_dasar

    def get_stok(self):
        return self.__stok

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {jumlah} unit.")

    def get_harga_dasar(self):
        return self.__harga_dasar

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 10 // 100
        print(f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,} | Pajak(10%): Rp {pajak:,}")

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 10 // 100
        harga_per_malam = self.get_harga_dasar() + pajak
        return harga_per_malam * jumlah_malam


class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 5 // 100
        print(f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,} | Pajak(5%): Rp {pajak:,}")

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 5 // 100
        harga_per_malam = self.get_harga_dasar() + pajak
        return harga_per_malam * jumlah_malam


def proses_transaksi(daftar_pesanan):
    total_tagihan = 0

    print("--- STRUK PEMESANAN ---")

    for i, pesanan in enumerate(daftar_pesanan, 1):
        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["malam"]

        print(f"{i}. ", end="")
        kamar.tampilkan_detail()

        subtotal = kamar.hitung_harga_total(jumlah_malam)

        print(f"Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,}")
        print()

        total_tagihan += subtotal

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,}")
    print("----------------------------------------")


# --- SETUP DATA KAMAR ---

print("--- SETUP DATA KAMAR ---")

deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1500000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500000,
    "2 Orang"
)

deluxe.tambah_stok(10)
standard.tambah_stok(-5)
standard.tambah_stok(20)

print()

# --- PEMESANAN ---

daftar_pesanan = [
    {
        "kamar": deluxe,
        "malam": 2
    },
    {
        "kamar": standard,
        "malam": 1
    }
]

proses_transaksi(daftar_pesanan)
```
