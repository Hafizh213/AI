```python
class Kucing:
    # Constructor
    def __init__(self, nama, warna):
        self.nama = nama  # Attribute
        self.warna = warna  # Attribute

    # Method
    def suara(self):
        return f"{self.nama} berkata: Opal!"

# Membuat Object (Instansiasi)
kucing1 = Kucing("Aryoo", "Putih")
kucing2 = Kucing("Bima", "Oranye")

# Mengakses Attribute dan Method
print(kucing1.nama)  # Output: Luna
print(kucing2.suara())  # Output: Garfield berkata: Meow!
```

```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name # Nama Hero
        self.hp = hp # Nyawa (Health Point)
        self.attack_power = attack_power # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

# Method menyerang: Objek ini (self) menyerang objek lain (lawan)
def serang(self, lawan):
    print(f"{self.name} menyerang {lawan.name}!")
    lawan.diserang(self.attack_power)
# Method diserang: Menerima damage
def diserang(self, damage):
    self.hp -= damage
# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

Tugas Analisis 2

Parameter lawan menerima sebuah objek agar method serang() dapat mengakses atribut dan method milik lawan, seperti lawan.name dan lawan.diserang(). Jika hanya menerima string nama, program hanya mengetahui nama hero dan tidak bisa mengurangi HP atau memanggil method diserang(). Dengan menerima objek utuh, interaksi antarobjek dapat dilakukan secara langsung.

```python
# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 10)
balmond = Hero("Balmond", 200, 10)
eudora.info()
eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0 # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(1001) # Coba set negatif
print(hero1.get_hp()) # Output: 0 (Karena dicegat oleh logika
# Setter)
```

```python



# Class Hero baru untuk demonstrasi Enkapsulasi
# Perhatian: Mendefinisikan ulang class Hero akan menimpa definisi sebelumnya.
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba Enkapsulasi --
hero_encapsulated = Hero("Layla", 100)
# hero_encapsulated.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero_encapsulated.__hp) # ERROR (Tidak bisa dibaca langsung)
hero_encapsulated.set_hp(-50) # Coba set negatif
print(hero_encapsulated.get_hp()) # Output: 0 (Karena dicegat oleh setter)

print(f"Mencoba akses paksa: {hero_encapsulated._Hero__hp}")



from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass
    @abstractmethod
    def info(self):
        pass
# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama
    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")
    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis
    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")
    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")
# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
m.serang("Serigala")
m.serang("Alucard")

#Praktek 6-----------#

# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama
    def serang(self):
        print("Hero menyerang dengan tangan kosong.")
# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")
# Child Class 2
class Archer(Hero):

    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")
# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]
print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```

Jika method serang() pada class Hero dihapus, saat menjalankan:

akan muncul error:

TypeError: Can't instantiate abstract class Hero with abstract method serang

Artinya, class Hero belum memenuhi kontrak dari class GameUnit. Karena GameUnit mewajibkan setiap turunannya memiliki method serang() dan info(), maka Hero harus membuat kedua method tersebut.

Konsekuensinya: objek Hero tidak dapat dibuat sebelum semua method abstract yang diwajibkan sudah dibuat.

2. Mencetak Cetakan

Jika baris:

diaktifkan, akan muncul error karena GameUnit adalah abstract class dan masih memiliki method abstract yang belum memiliki implementasi.

GameUnit tidak dibuat untuk menjadi objek nyata, tetapi sebagai cetakan atau kontrak bagi class turunannya seperti Hero dan Monster.

Gunanya adalah memastikan semua GameUnit memiliki method yang sama, yaitu:

serang()
info()

Dengan begitu, setiap class yang mewarisi GameUnit wajib menyediakan fungsi tersebut.

Karena polimorfisme bekerja dengan menggunakan method yang sama (serang) tetapi memiliki perilaku yang berbeda pada setiap child class. Jika namanya berbeda, Python menganggapnya sebagai method yang berbeda sehingga tidak terjadi overriding.
